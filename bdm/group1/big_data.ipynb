```python
# Declaring File Path
import os

# The exact path based on your folder and file name
file_path = '/content/drive/MyDrive/Shopee Dataset (HPDP Asm 2)/delivery_orders_march.csv'

# Quick check to make sure the file is visible
if os.path.exists(file_path):
    print("File found! Ready to proceed.")
else:
    print("File not found. Please double-check the folder name in your Drive.")
```

```python
# Performing Dataset with Pandas Library
import pandas as pd
```

```python
# Checking Metadata
import time

# 1. Initial Load (Baseline)
start_time = time.time()
df = pd.read_csv(file_path)
end_time = time.time()

#  2. Inspection
print(f"Time Taken to Load: {end_time - start_time:.2f} seconds")
print(f"Dataset Shape: {df.shape}") # Shows total rows and columns
print("\n--- Column Data Types ---")
print(df.dtypes) # Requirement: Report data types
print("\n--- Missing Values ---")
print(df.isnull().sum()) # Requirement: Summary of missing values
```

```python
# Previewing Dataset
df_preview = pd.read_csv(file_path, nrows=5)
display(df_preview)
```

```python
# Transforming Datatype
# Convert the '1st_deliver_attempt' column to a readable date
df['1st_deliver_attempt'] = pd.to_datetime(df['1st_deliver_attempt'], unit='s')
print(df['1st_deliver_attempt'].head())
```

```python
# Convert the '2nd_deliver_attempt' column to a readable date
df['2nd_deliver_attempt'] = pd.to_datetime(df['2nd_deliver_attempt'], unit='s')
print(df['2nd_deliver_attempt'].head())
```

```python
# Big Data Handling Strategies
# Strategy 1: Load Less Data
# Only load specific columns to save RAM immediately
cols = ['orderid', 'pick', '1st_deliver_attempt', '2nd_deliver_attempt']
df_less = pd.read_csv(file_path, usecols=cols)
print(f"Memory with all columns: {df.memory_usage().sum() / 1e6:.2f} MB")
print(f"Memory with 'Load Less' strategy: {df_less.memory_usage().sum() / 1e6:.2f} MB")
```

```python
# Strategy 2: Chunking
# Reading in chunks of 100,000 rows at a time[cite: 1]
chunk_size = 100000
batch_no = 1

for chunk in pd.read_csv(file_path, chunksize=chunk_size):
    # Example: Perform a simple count or filter on each chunk
    print(f"Processing Chunk #{batch_no}: {len(chunk)} rows")
    batch_no += 1
```

```python
# Strategy 3: Data Type Optimisation
# Before: Check memory of the specific column
print(f"Memory Before: {df_less['1st_deliver_attempt'].memory_usage() / 1e6:.2f} MB")

# Optimization: Convert to Datetime[cite: 1]
df_less['1st_deliver_attempt'] = pd.to_datetime(df_less['1st_deliver_attempt'], unit='s')

# Optimization: Convert orderid to string or category if it saves space[cite: 1]
df_less['orderid'] = df_less['orderid'].astype('category')

print(f"Memory After Optimization: {df_less.memory_usage().sum() / 1e6:.2f} MB")
```
